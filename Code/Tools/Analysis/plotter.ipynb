```python
import pyBrellaSampling.Code.Tools.io as io
import pyBrellaSampling.Code.Tools.visualisation as vis
import numpy
import os
from pprint import pprint
```

```python
qmmm_dir = "/big/Q4Bio/Data/Test_QMMM/Water/Orca_2000/WHAM/metadata/"
qmmm_file = "wham.pmf"
qmmm_data = io.textRead(os.path.join(qmmm_dir,qmmm_file))
```

```python
qmmm_dict = {float(key.split()[0]):{"energy": key.split()[1],
                             "err": key.split()[2]} for key in qmmm_data[1:] if "#" not in key}
qmmm_x = qmmm_dict.keys()
qmmm_en = [float(qmmm_dict[x]["energy"]) for x in qmmm_x]
```

```python
def zero_min(dat:list):
    min = numpy.min(dat)
    return [i - min for i in dat]
```

```python
qm_loc = "/big/Q4Bio/Data/Test_QMMM/Water/GasPhase/no_solvent/"
qm_file = "Orca.relaxscanact.dat"

qm_data = io.textRead(os.path.join(qm_loc, qm_file))
qm_x = [float(dat.split()[0]) for dat in qm_data]
qm_y = [float(dat.split()[1])*627.5 for dat in qm_data]

qm_solv_loc = "/big/Q4Bio/Data/Test_QMMM/Water/GasPhase/implicit_solvent"
qm_solv_data = io.textRead(os.path.join(qm_solv_loc, qm_file))
qm_solv_x = [float(dat.split()[0]) for dat in qm_solv_data]
qm_solv_y = [float(dat.split()[1])*627.5 for dat in qm_solv_data]
```

```python
plot = vis.Plot2D(qm_x, [zero_min(qm_y), zero_min(qm_solv_y)], ["Gas Phase", "Implicit Water"], "", "O-H bond distance (Å)", "Energy (Kcal/mol)")
```

```python
plot.add_scatter(x=list(qmmm_x), y=zero_min(qmmm_en), name="QM/MM", marker={"color":vis.PlotColourList[2].to_string()})
```

```python
plot.update_layout(template="simple_white")
plot.show()
```
